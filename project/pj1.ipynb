```python
df = pd.read_csv('D2.csv', encoding='cp949')

df.head()

monthly = (
    df.groupby('연월')['물동량']
    .sum()
    .reset_index()
)

monthly
```

```python
monthly_avg = (
    monthly.groupby('월')['물동량']
           .mean()
           .reset_index()
)

monthly_avg 
```

```python
avg = monthly_avg['물동량'].mean()

monthly_avg['평균차이'] = monthly_avg['물동량'] - avg

monthly_avg
```

```python
monthly_avg['차이율(%)'] = (
    monthly_avg['평균차이'] / avg * 100
).round(2)

monthly_avg
```

부산항의 물동량의 차이를 보면 2월달에 현저히 물동량이 떨어지는것을 볼수 있다. 이후 점차 회복하다가 다시 감소한다.

환적물량의 성장세는 수출입량의 성장세보다 높은것을 보았을때 부산항은 환적황으로서의 발전가능성이 높다는것을 알 수 있다.
